```python
import re
import pandas as pd
import matplotlib.pyplot as plt
import unicodedata
from pathlib import Path

# Busca la carpeta "data" subiendo desde donde se ejecuta el notebook
# (funciona igual desde la raíz del proyecto, rules/ o distributions/)
DATA = next(p / "data" for p in [Path.cwd(), *Path.cwd().parents] if (p / "data").is_dir())

RUTA = DATA / "FORMULARIO DE CONTEXTO limpio.csv"

df = pd.read_csv(RUTA, encoding="utf-8-sig")
cantidad_mujeres = (df["Sexo"] == "Femenino").sum()
cantidad_hombres = (df["Sexo"] == "Masculino").sum()
camtidad_otros = (df["Sexo"] == "Otro").sum()

etiquetas = ["Mujeres", "Hombres", "Otros"]
valores = [cantidad_mujeres, cantidad_hombres, camtidad_otros]

plt.pie(valores, labels=etiquetas, autopct="%1.1f%%", startangle=90)
plt.title("Distribución de Género")
plt.show()



```
