Limpieza de Datos: Columna "Semestre"

En este apartado se estandarizó la columna de Semestre. El principal reto fue la falta de restricciones en la encuesta, lo que generó una alta variabilidad en las respuestas de los alumnos.

Problema identificado: Heterogeneidad en los formatos de entrada.

Formatos detectados: "Segundo Semestre", "2do", "2".

```python
df = pd.read_csv("data/FORMULARIO DE CONTEXTO.csv")


"""Vamos a extraer en este apartado uno o mas digitos por respuesta en el semestre"""
df['semestre_limpio'] = df['Semestre'].astype(str).str.extract(r'(\d+)')

# 3. Convertimos ese numero a texto
df['semestre_limpio'] = df['semestre_limpio'].astype('Int64')

print(df[['Semestre', "semestre_limpio"]].head(20))

"""Aqui vamos a renombrar la tabla limpia para que se llame a la original"""

df = df.drop(columns=["Semestre"])
df = df.rename(columns={"semestre_limpio" : 'Semestre'})

# SObreescribimos en el csv
df.to_csv("data/FORMULARIO DE CONTEXTO.csv", index=False, encoding="utf-8")

print("Archivo actualizado con exito")
```

```python

import pandas as pd

ruta = "data/FORMULARIO DE CONTEXTO.csv"
df = pd.read_csv(ruta, encoding="utf-8-sig")

# 1. Limpiar los espacios de todos los encabezados
df.columns = df.columns.str.strip()

# 2. Normalizar las respuestas de semestre
semestre = (
    df["Semestre"]
    .astype("string")
    .str.lower()
    .str.strip()
)

# 3. Traducir respuestas escritas a números
traducciones = {
    "primero": "1",
    "primer semestre": "1",
    "1ro": "1",
    "1er": "1",
    "1er semestre": "1",
    "1°": "1",
    "1° semestre": "1",
    "segundo": "2",
    "segundo semestre": "2",
    "2do": "2",
    "2do semestre": "2",
    "tercero": "3",
    "tercer semestre": "3",
    "3ro": "3",
    "3er": "3",
    "cuarto": "4",
    "cuarto semestre": "4",
    "4to": "4",
    "quinto": "5",
    "quinto semestre": "5",
    "sexto": "6",
    "sexto semestre": "6",
    "septimo": "7",
    "séptimo": "7",
    "séptimo semestre": "7",
    "7mo": "7",
    "octavo": "8",
    "octavo semestre": "8",
    "noveno": "9",
    "decimo": "10",
    "décimo": "10",
}

semestre = semestre.replace(traducciones)

# 4. Convertir a enteros, conservando faltantes como <NA>
df["Semestre"] = pd.to_numeric(
    semestre, errors="coerce"
).astype("Int64")

# 5. Revisar los resultados antes de guardar
print(df["Semestre"].value_counts(dropna=False).sort_index())

# 6. Guardar el CSV actualizado
df.to_csv(ruta, index=False, encoding="utf-8-sig")

print("CSV actualizado correctamente.")
```
