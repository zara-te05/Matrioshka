Reglas para la encuesta en el apartado de Capital 

```python
# REGLAS PARTICULARES DEL HOGAR

import pandas as pd
# ESTA ES LA LÍNEA QUE CREA LA VARIABLE "df"
df = pd.read_csv("data/FORMULARIO DE CONTEXTO.csv")

regla_tamano_de_la_vivienda = {
    "Menos de 50 m²: Corresponde a departamentos pequeños o viviendas de interés social" : 0.0,
    "Entre 50 y 80 m²: Es el estándar de muchas casas de fraccionamientos o viviendas urbanas promedio" : 0.25, 
    "Entre 81 y 120 m²: Viviendas de tamaño medio con mayor número de habitaciones" : 0.50,
    "Entre 121 y 200 m²: Viviendas amplias, usualmente de dos niveles o con terrenos extensos" : 0.75,
    "Más de 200 m²: Residencias de grandes dimensiones o propiedades con terrenos amplios." : 1.0
}

regla_personas_en_casa = {
    "1" : 0,
    "De 2 a 3" : 0.25,
    "De 3 a 4" : 0.50,
    "De 4 a 5" : 0.75,
    "Mas de 5" : 1
}

regla_de_la_habitacion = {
    "Solo yo" : 1,
    "Dos personas" : 0.75,
    "Entre dos y tres personas" : 0.50,
    "Entre tres y cuatro personas" : 0.25,
    "Cinco o mas" : 0
}

regla_de_la_liquidez = {
    'Muy facil' : 0.0,
    'Facil' : 0.25,
    'Moderado' : 0.50,
    'Dificil' : 0.75,
    'Muy dificil' : 1.0 
    
}

regla_promedio_ingresos_mensuales =  {
    'Menos de $7,500 MXN' : 1.0,
    'De $7,501 a $15,000 MXN' : 0.75,
    'De $15,001 a $22,500 MXN' : 0.50,
    'De $22,501 a $37,500 MXN' : 0.25,
    'Mas de $37,500 MXN' : 0.0
}

regla_del_tiempo_de_transporte = {
    'Menos de 30 minutos' : 0.0,
    'Entre 30 minutos y 1 hora' : 0.25,
    'Entre 1 hora y 1 hora 30 minutos' : 0.50,
    'Entre 1 hora 30 minutos y 2 horas' : 0.75,
    'Mas de 2 horas diarias' : 1.0
}

regla_costos_transporte_por_semana = {
    'Menos de $150 MXN a la semana.' : 0.0,
    'Entre $250 y $400 MXN a la semana.' : 0.25,
    'Entre $450 y $550 MXN a la semana.' : 0.50,
    'Entre de $600 a $750 MXN a la semana.' : 0.75,
    'Mas de $750 MXN a la semana' : 1.0
}

# Aplicamos aqui la conversion a numeros

df["Tamano casa"] = df["Tamano casa"].astype(str).str.strip().str.rstrip('., ').replace(regla_tamano_de_la_vivienda).astype(float)
df["Personas casa"] = df["Personas casa"].astype(str).str.strip().str.rstrip('., ').replace(regla_personas_en_casa).astype(float)
df['Habitaciones'] = df['Habitaciones'].astype(str).str.strip().str.rstrip('., ').replace(regla_de_la_habitacion).astype(float) # <-- AQUÍ ESTÁ EL CAMBIO
df['Dificultad liquidez'] = df['Dificultad liquidez'].astype(str).str.rstrip('., ').replace(regla_de_la_liquidez).astype(float)
df['Ingreso mensual'] = df['Ingreso mensual'].astype(str).str.rstrip('., ').replace(regla_promedio_ingresos_mensuales).astype(float)
df['Tiempo transporte'] = df['Tiempo transporte'].astype(str).str.rstrip('., ').replace(regla_del_tiempo_de_transporte).astype(float)


#----------------
# Paso A: Vamos a calcular el "Espacio disponible" 
Espacio = (df['Tamano casa'] + df['Habitaciones']) / 2

# Paso B: Calculamos el Hacinamiento
Hacinamiento = df['Personas casa'] # Te corregí el error de tipeo "Hacimiento" -> "Hacinamiento"

# Paso C: Formula Final.
df['Indice_Confort_Hogar'] = (Espacio - Hacinamiento + 1) / 2
df['Indice_Confort_Hogar'] = df['Indice_Confort_Hogar'].round(2)

print(df[['Tamano casa', 'Personas casa','Habitaciones','Indice_Confort_Hogar']].head(10))
# Esto te mostrará la lista exacta de nombres de tu CSV
print(df.columns.tolist())

# Lista de columnas corregida con comas en su lugar
columnas_a_guardar = [
    'Nombre Completo del Alumno', 'Agua', 'Luz', 
    'Internet', 'Equipo propio', 'Equipo lento', 
    'Banos', 'Tamano casa', 'Personas casa', 
    'Habitaciones', 'Seguridad transporte', 'Costos transporte', 
    'Tiempo transporte', 'Ingreso mensual', 'Dificultad liquidez', 
    'Indice_Confort_Hogar'
]

# Filtramos asegurándonos de que no truene si hay diferencias de texto menores
df[columnas_a_guardar].to_csv("data/CONDICIONES MATERIALES.csv", index=False)

print("¡Archivo guardado exitosamente!")


print("¡Archivo guardado!") 
```
